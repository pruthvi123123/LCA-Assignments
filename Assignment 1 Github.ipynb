```python
# 1. Setup collections (Tuple for labels, List for tracking keys, Dict for database)
labels = ("Roll Number", "Name", "Branch", "Marks")
roll_list = [101, 102]
students = {
    101: {"Name": "Alice", "Branch": "CSE", "Marks": 85},
    102: {"Name": "Bob", "Branch": "ECE", "Marks": 78}
}

print(f"Categories (Tuple): {labels}")
print(f"Initial Records: {students}\n")

# --- PERFORMING OPERATIONS ---

# A. ADD a new student record
new_roll = 103
if new_roll not in students:
    roll_list.append(new_roll)
    students[new_roll] = {"Name": "Charlie", "Branch": "ME", "Marks": 90}
    print(f"Added: Roll {new_roll}")

# B. UPDATE details of an existing student
up_roll = 101
if up_roll in students:
    students[up_roll]["Marks"] = 95
    print(f"Updated: Roll {up_roll} marks to 95")

# C. DELETE an existing student record
del_roll = 102
if del_roll in students:
    del students[del_roll]
    roll_list.remove(del_roll)
    print(f"Deleted: Roll {del_roll}")

# D. DISPLAY final student records
print("\n--- Final Student Records ---")
for roll in roll_list:
    if roll in students:
        info = students[roll]
        print(f"Roll: {roll} | Name: {info['Name']} | Branch: {info['Branch']} | Marks: {info['Marks']}")
```
